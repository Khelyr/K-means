```python
def negative_correction(x, minimum):
    if minimum <0:
        return float(x - (minimum-1))
    

def no_notypes(x):
    if type(x) == "NoneType":
        return int(x)
```

```python
minimum = dataset1.min()
dataset1[0] = dataset1[0].apply(lambda x: negative_correction(x, minimum[0]))
dataset1[1] = dataset1[1].apply(lambda x: negative_correction(x, minimum[0]))
dataset1[0] = dataset1[0].apply(lambda x: round(float(x), 2))
dataset1[1] = dataset1[1].apply(lambda x: round(float(x), 2))
dataset1
```
