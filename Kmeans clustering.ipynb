```python
#импортируем библиотеки...
import pandas as pd #библиотека для эффективной работы с данными
import numpy as np #библиотека для работы с наборами данных
import matplotlib.pyplot as plt #библиотека для визализации
import seaborn as sns #библиотека для построения графиков
from mpl_toolkits.mplot3d import axes3d
```

```python
%matplotlib notebook
```

```python
dataset1 = pd.read_csv("D:\Projects\CPP\Italy\\rawdata\dataset1.csv", header = None, sep = "\s+")
dataset1
```

```python
plt.figure()
plt.scatter(dataset1[0], dataset1[1])
```

```python
def negative_correction(x, minimum):
    if minimum <0:
        return x - minimum
```

```python
minimum = dataset1.min()
dataset1[0] = dataset1[0].apply(lambda x: negative_correction(x, minimum[0]-1))
dataset1[1] = dataset1[1].apply(lambda x: negative_correction(x, minimum[0]-1))
dataset1[0] = dataset1[0].apply(lambda x: round(float(x), 2))
dataset1[1] = dataset1[1].apply(lambda x: round(float(x), 2))
dataset1
```

```python
plt.figure()
plt.scatter(dataset1[0], dataset1[1])
```

```python
dataset1.to_csv("D:\Projects\CPP\Italy\dataset1.csv", header = None, index = False, sep = " ")
```

```python
dataset1_out5clusters = pd.read_csv("D:\Projects\CPP\Italy\dataset1_out5clusters.csv", header = None, sep = "\s+")
dataset1_out5clusters
```

```python
clusters_amount = dataset1_out5clusters.max()[0]
dataset1_out5clusters.sort_values(0, ascending=True)
```

```python
plt.figure()
clusters_amount = dataset1_out5clusters.max()[0]
for i in range(int(clusters_amount)+1):
    cluster = []
    for j in range(dataset1_out5clusters.shape[0]):
        if int(dataset1_out5clusters[0][j]) == i:
            cluster.append(j)
    coord_x = [dataset1_out5clusters[1][k] for k in cluster]
    coord_y = [dataset1_out5clusters[2][k] for k in cluster]
    plt.scatter(coord_x, coord_y)
```
